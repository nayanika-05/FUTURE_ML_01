```python
# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# %%
df = pd.read_csv('../data/superstore.csv', encoding='latin1')

df.head()

# %%
df = pd.read_csv('../data/SampleSuperstore.csv', encoding='latin1')

df.head()

# %%
import pandas as pd

# %%
df.info()

# %%
df.describe()

# %%
df.isnull().sum()

# %%
df = df.drop_duplicates()

print("Duplicates removed successfully")

# %%
df['Order Date'] = pd.to_datetime(df['Order Date'])

# %%
df.info()

# %%
df.head()

# %%
df.info()

# %%
df.describe()

# %%
df.isnull().sum()

# %%
df = df.drop_duplicates()

print("Duplicates removed successfully")

# %%
df.info()

# %%
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Day'] = df['Order Date'].dt.day
df['Weekday'] = df['Order Date'].dt.weekday
df['Quarter'] = df['Order Date'].dt.quarter

# %%
df.head()

# %%
daily_sales = df.groupby('Order Date')['Sales'].sum().reset_index()

# %%
daily_sales.head()

# %%
plt.figure(figsize=(12,6))

plt.plot(
    daily_sales['Order Date'],
    daily_sales['Sales']
)

plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.savefig('../images/sales_trend.png')
plt.show()

# %%
daily_sales['Day_Number'] = np.arange(len(daily_sales))

# %%
daily_sales.head()

# %%
X = daily_sales[['Day_Number']]

# %%
y = daily_sales['Sales']

# %%
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    shuffle=False
)

# %%
model = RandomForestRegressor()

# %%
model.fit(X_train, y_train)

# %%
predictions = model.predict(X_test)

# %%
mae = mean_absolute_error(y_test, predictions)

print("MAE:", mae)

# %%
rmse = np.sqrt(mean_squared_error(y_test, predictions))

print("RMSE:", rmse)

# %%
plt.figure(figsize=(12,6))

plt.plot(y_test.values, label='Actual Sales')
plt.plot(predictions, label='Predicted Sales')

plt.legend()

plt.title('Actual vs Predicted Sales')
plt.savefig('../images/prediction_comparison.png')
plt.show()

# %%
future_days = pd.DataFrame({
    'Day_Number':
    np.arange(
        len(daily_sales),
        len(daily_sales) + 30
    )
})

# %%
future_predictions = model.predict(future_days)

# %%
plt.figure(figsize=(12,6))

plt.plot(future_predictions)

plt.title('Next 30 Days Sales Forecast')
plt.xlabel('Future Days')
plt.ylabel('Predicted Sales')
plt.savefig('../images/future_forecast.png')
plt.show()

# %%



```
